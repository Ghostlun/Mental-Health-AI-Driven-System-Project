```python
from datasets import load_dataset

dataset = load_dataset("daekeun-ml/naver-news-summarization-ko")
data = dataset
data
```

```python
# Data Test
data["train"]["document"][0]  # Accessing the first document in the training set
print(data["train"]["document"][0])
ko_text = "".join(data["train"]["document"])
ko_chars = sorted(list(set(ko_text)))
ko_vocab_size = len(ko_chars)
print(f"Number of unique characters in the training set: {ko_vocab_size}")
print(ko_chars)
```

```python
chracter_to_ids = {char: i for i, char in enumerate(ko_chars)}
ids_to_character = {i: char for i, char in enumerate(ko_chars)}
token_encode =  lambda s:[chracter_to_ids[c] for c in s]
token_decode = lambda l: "".join([ids_to_character[i] for i in l])

encoded = token_encode("안녕하세요 인공지능 공부하게 되어 반가워요")
decoded = token_decode(encoded)

print("Encoded:", encoded)
print("Decoded", decoded)
```

```python
import torch

# Tokenization and Tensor Conversion
tokennized_data = torch.tensor(token_encode(ko_text), dtype=torch.long)
print(tokennized_data.shape, tokennized_data.dtype)
print(tokennized_data[:100])
```

```python
# Data type conversion
n = int(0.9 * len(tokennized_data))
# Train Dataset 90%
train_dataset = tokennized_data[:n]
# Test Dataset 10%
test_dataset = tokennized_data[n:]
```

✅ batch
Definition: The number of samples (e.g., text inputs) processed by the model at once.
Analogy: Like how many students take a test at the same time.

✅ block_size
Definition: The maximum length of text (in tokens) the model can handle at one time.
Analogy: Like how many pages you read in one sitting.

```python
block_size = 8 # length of tensor
batch_size = 0
x = train_dataset[:block_size]
y = train_dataset[block_size:]

for time in range(block_size):
    context = x[:time+1]
    target = y[time]

    print(f"Input tensor : {context}")
    print(f"Target {target}")
```

```python
# Batch size

torch.manual_seed(1234)

batch_size = 4
block_size = 8

def batch_function(mode):
    dataset = train_dataset if mode == "train" else test_dataset
    idx = torch.randint(len(dataset) - block_size, (batch_size,))
    x = torch.stack([dataset[index: index + block_size] for index in idx])
    y = torch.stack([dataset[index + 1: index+ block_size + 1] for index in idx])
    return x, y

example_x, exmaple_y = batch_function("train")
print("inputs :", example_x.shape)
print("")
print("example value")
print(example_x)
print("-------------")
print("target : ", exmaple_y.shape)
print("")


for size in range(batch_size):
    for t in range(block_size):
        context = example_x[size, : t+1]
        target = exmaple_y[size, t]
        print(f"input : {context}, target: {target}")
    
    print("-------------")
    print("-------------")
```

```python
import torch
import torch.nn as nn
from torch.nn import functional as F

# Loss Added Handled version
# More precisely, it’s because your class inherits from torch.nn.Module.
# PyTorch’s nn.Module includes the automatic calling logic where __call__() internally triggers the forward() method.
class semiGPT(nn.Module):
    def __init__(self, vocab_length):
        super().__init__()
        self.embedding_token_table = nn.Embedding(vocab_length, vocab_length)
    
    def forward(self, inputs, targets):
        logits = self.embedding_token_table(inputs)
        batch, seq_length, vocab_length = logits.shape
        logits = logits.view(batch * seq_length, vocab_length)
        targets = targets.view(batch * seq_length)
        loss = F.cross_entropy(logits, targets)

        print(f"logits's shape: {logits.shape}")
        print(f"target's shape: {targets.shape}")
        return logits, loss

model = semiGPT(ko_vocab_size)
logits, loss = model(exmaple_y, exmaple_y)
print(loss)
```

```python
# Generate Method
import torch
import torch.nn as nn
from torch.nn import functional as F

# Loss Added Handled version
# logits are the raw, unnormalized scores that the model outputs for each possible token in 
# its vocabulary before they are converted into probabilities
class semiGPT(nn.Module):
    def __init__(self, vocab_length):
        super().__init__()
        self.embedding_token_table = nn.Embedding(vocab_length, vocab_length)
    
    def forward(self, inputs, targets=None):
        logits = self.embedding_token_table(inputs)
        if targets is None:
            loss = None
        else:
            batch, seq_length, vocab_length = logits.shape
            logits = logits.view(batch * seq_length, vocab_length)
            targets = targets.view(batch*seq_length)
            loss = F.cross_entropy(logits, targets)
        return logits, loss

    
    def generate(self, inputs, max_new_tokens):
        for _ in range(max_new_tokens):
            logits, _ = self.forward(inputs)             # (B, T, vocab)
            logits = logits[:, -1, :]                    # (B, vocab)
            probs = F.softmax(logits, dim=-1)            # (B, vocab)
            next_token = torch.multinomial(probs, num_samples=1)  # (B, 1)
            inputs = torch.cat((inputs, next_token), dim=1)       # (B, T+1)
        return inputs
    
    

model = semiGPT(ko_vocab_size)
logits, loss = model(exmaple_y, exmaple_y)
print(loss)

token_decode(model.generate(torch.zeros((1,1),
                            dtype = torch.long),
                            max_new_tokens=10)[0].tolist())
```

```python
import torch.optim


learning_rate = 1e-2
# Adding Optimizing

model = semiGPT(ko_vocab_size)
optimizer = torch.optim.AdamW(model.parameters(), lr= learning_rate)


from tqdm.auto import tqdm
batch_size = 32

for steps in tqdm(range(10000)):
    example_x, exmaple_y = batch_function("train")
    logits, loss = model(example_x, exmaple_y)
    optimizer.zero_grad(set_to_none=True)
    loss.backward()
    optimizer.step()

print(loss.item())
```
